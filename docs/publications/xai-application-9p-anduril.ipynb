## 1. Gauge Constraint — The Conservation Law

Every operation in the system must satisfy `α + ω = 15`. This is the architectural invariant — not a preference, not a heuristic. The optimal point is the **Viviani peak** at `α=7, ω=8`, where the Fibonacci-weighted product maximizes.

This cell implements the full gauge enforcement layer with error types, scoring, and Fibonacci weighting.

````python
// ── Gauge Constraint: α + ω = 15 ──────────────────────────────────
// Every pipeline stage, every 9P transaction, every ATOM trail entry
// must satisfy this invariant. No exceptions.

const PHI: f64 = 1.618033988749895;
const ALPHA_OMEGA_TARGET: u32 = 15;
const GAUGE_TOLERANCE: f64 = 0.00055;

#[derive(Debug)]
enum ConservationError {
    Violation { alpha: u32, omega: u32, sum: u32, delta: f64 },
    AlphaOutOfRange(u32),
    OmegaOutOfRange(u32),
}

impl std::fmt::Display for ConservationError {
    fn fmt(&self, f: &mut std::fmt::Formatter) -> std::fmt::Result {
        match self {
            Self::Violation { alpha, omega, sum, delta } =>
                write!(f, "Gauge violation: α({}) + ω({}) = {}, expected {} (δ={})",
                    alpha, omega, sum, ALPHA_OMEGA_TARGET, delta),
            Self::AlphaOutOfRange(v) => write!(f, "Alpha out of range [0, 15]: {}", v),
            Self::OmegaOutOfRange(v) => write!(f, "Omega out of range [0, 15]: {}", v),
        }
    }
}

#[derive(Debug, Clone)]
struct GaugeScore {
    alpha: u32,
    omega: u32,
    sum: u32,
    delta: f64,
    viviani_resonance: bool,
}

struct GaugeConstraint { alpha: u32, omega: u32 }

impl GaugeConstraint {
    fn new(alpha: u32, omega: u32) -> Self { Self { alpha, omega } }

    fn verify(&self) -> Result<GaugeScore, ConservationError> {
        if self.alpha > 15 { return Err(ConservationError::AlphaOutOfRange(self.alpha)); }
        if self.omega > 15 { return Err(ConservationError::OmegaOutOfRange(self.omega)); }
        let sum = self.alpha + self.omega;
        let delta = (sum as f64 - ALPHA_OMEGA_TARGET as f64).abs();
        if delta > GAUGE_TOLERANCE {
            return Err(ConservationError::Violation {
                alpha: self.alpha, omega: self.omega, sum, delta,
            });
        }
        Ok(GaugeScore {
            alpha: self.alpha, omega: self.omega, sum, delta,
            viviani_resonance: self.alpha == 7 && self.omega == 8,
        })
    }

    fn fibonacci_score(&self) -> f64 {
        let fib: [f64; 16] = [
            1.0, 1.0, 2.0, 3.0, 5.0, 8.0, 13.0, 21.0,
            34.0, 55.0, 89.0, 144.0, 233.0, 377.0, 610.0, 987.0,
        ];
        let product = fib[self.alpha as usize] * fib[self.omega as usize];
        (product / 714.0).min(1.0) // 714 = fib(7) * fib(8) = Viviani peak
    }
}

// ── Demonstrate ───────────────────────────────────────────────────

println!("=== Gauge Constraint: α + ω = {} ===", ALPHA_OMEGA_TARGET);
println!();

// Test all valid pairs
println!("Valid pairs (all must pass):");
for alpha in 0..=15u32 {
    let omega = 15 - alpha;
    let g = GaugeConstraint::new(alpha, omega);
    let score = g.verify().unwrap();
    let fib = g.fibonacci_score();
    let marker = if score.viviani_resonance { " ← VIVIANI PEAK" } else { "" };
    println!("  α={:2}, ω={:2}  fib_score={:.4}{}", alpha, omega, fib, marker);
}

println!();

// Test violation
let bad = GaugeConstraint::new(7, 7);
match bad.verify() {
    Err(e) => println!("Violation correctly caught: {}", e),
    Ok(_) => println!("ERROR: Should have rejected α=7, ω=7"),
}
````

## 2. Fibonacci Anyon Fusion — τ ⊗ τ = 1 ⊕ τ

The engine implements SU(2)₃ Fibonacci anyons — the simplest non-Abelian anyon model that achieves universal quantum computation through braiding alone.

Three lattice types encode the computation:
- **Θ Lattice** (3 qubits): Fusion logic via the F-matrix
- **Tetrahedron** (6 qubits): Chiral fusion / Racah-Wigner 6j symbols
- **Tailed Θ** (9 qubits): Universal braiding with Qv correction (BQP-complete)

The conservation law emerges from the physics: **Tailed Θ (9) + Tetrahedron (6) = 15**.

````python
// ── Fibonacci Anyon Fusion & Lattice Types ──────────────────────────

#[derive(Debug, Clone, Copy, PartialEq)]
enum Anyon { Vacuum, Tau }

#[derive(Debug)]
struct FusionResult { channel: Anyon, amplitude: f64 }

fn fuse(a: Anyon, b: Anyon) -> Vec<FusionResult> {
    match (a, b) {
        (Anyon::Vacuum, x) | (x, Anyon::Vacuum) =>
            vec![FusionResult { channel: x, amplitude: 1.0 }],
        (Anyon::Tau, Anyon::Tau) => vec![
            FusionResult { channel: Anyon::Vacuum, amplitude: 1.0 / (PHI * PHI) },
            FusionResult { channel: Anyon::Tau, amplitude: 1.0 / PHI },
        ],
    }
}

// Θ Lattice: 3 qubits — encodes the F-matrix
struct ThetaLattice;
impl ThetaLattice {
    fn evaluate() -> f64 { 1.0 / PHI } // Θ(τ,τ,τ) = φ⁻¹
}

// Tetrahedron: 6 qubits — chiral fusion
struct TetrahedronLattice;
impl TetrahedronLattice {
    fn evaluate() -> f64 { 1.0 / (PHI * PHI) } // Tet(τ⁶) = φ⁻²
    fn chirality_index() -> f64 {
        let fwd = Self::evaluate();
        (fwd - 1.0 / fwd).abs()
    }
}

// Tailed Θ: 9 qubits — universal braiding (BQP-complete)
struct TailedThetaLattice;
impl TailedThetaLattice {
    fn qubit_count() -> usize { 9 }
    fn evaluate_with_qv() -> f64 {
        let base = ThetaLattice::evaluate();
        let qv_correction = PHI.powf(-1.5); // 3 vertices × φ^(-1/2)
        base * qv_correction
    }
}

// ── Demonstrate ───────────────────────────────────────────────────

println!("=== Fibonacci Anyon Fusion: τ ⊗ τ = 1 ⊕ τ ===");
println!();

let results = fuse(Anyon::Tau, Anyon::Tau);
for r in &results {
    println!("  Channel {:?}: amplitude = {:.6} (probability = {:.6})",
        r.channel, r.amplitude, r.amplitude);
}
println!("  Sum of amplitudes: {:.6} (= 1/φ² + 1/φ = 1.0)",
    results.iter().map(|r| r.amplitude).sum::<f64>());

println!();
println!("=== Lattice Evaluations ===");
println!("  Θ Lattice (3 qubits):       {:.6}", ThetaLattice::evaluate());
println!("  Tetrahedron (6 qubits):      {:.6}", TetrahedronLattice::evaluate());
println!("  Chirality index:             {:.6}", TetrahedronLattice::chirality_index());
println!("  Tailed Θ (9 qubits, +Qv):   {:.6}", TailedThetaLattice::evaluate_with_qv());
println!("  Expected: φ^(-5/2) =         {:.6}", PHI.powf(-2.5));

println!();
println!("=== The Conservation Law Emerges ===");
println!("  Tailed Θ qubits (α): {}", TailedThetaLattice::qubit_count());
println!("  Tetrahedron qubits (ω): 6");
println!("  α + ω = {} + 6 = {}",
    TailedThetaLattice::qubit_count(),
    TailedThetaLattice::qubit_count() + 6);
assert_eq!(TailedThetaLattice::qubit_count() as u32 + 6, 15);
println!("  ✓ Conservation holds.");
````

## 3. Braiding State Machine — B = F⁻¹ R F

The universal elementary braiding operator decomposes into three discrete states:

1. **F⁻¹** — Inverse F-matrix (change of basis into fusion space)
2. **R** — R-matrix phase accumulation, locked to a 20 Hz hardware clock stepping 8 ticks through a 16-hopper ring counter
3. **F** — Return to computational basis

R-matrix phases for Fibonacci anyons:
- R¹ₜₜ = e^{-4πi/5} (vacuum channel)
- Rτₜₜ = e^{3πi/5} (τ channel)

This eliminates continuous-phase drift entirely — every braid is a discrete sequence of exactly 10 ticks (500ms at 20Hz).

````python
// ── Braiding State Machine: B = F⁻¹ R F ──────────────────────────

use std::f64::consts::PI;

const R_VACUUM_PHASE: f64 = -4.0 * PI / 5.0;
const R_TAU_PHASE: f64 = 3.0 * PI / 5.0;
const TICKS_PER_BRAID: u8 = 8;
const RING_SIZE: u8 = 16;

#[derive(Debug, Clone, Copy, PartialEq)]
enum BraidState { InverseFMove, RMatrix, FMove, Complete }

#[derive(Debug)]
struct BraidOperator {
    state: BraidState,
    ring_position: u8,
    ticks_elapsed: u8,
    vacuum_amplitude: f64,
    tau_amplitude: f64,
    accumulated_phase: f64,
    total_braids_executed: u64,
}

impl BraidOperator {
    fn at_viviani() -> Self {
        Self {
            state: BraidState::InverseFMove,
            ring_position: 0,
            ticks_elapsed: 0,
            vacuum_amplitude: (1.0 / (PHI * PHI)).sqrt(),
            tau_amplitude: (1.0 / PHI).sqrt(),
            accumulated_phase: 0.0,
            total_braids_executed: 0,
        }
    }

    fn tick(&mut self) -> bool {
        match self.state {
            BraidState::InverseFMove => {
                let phi_inv = 1.0 / PHI;
                let phi_inv_sqrt = phi_inv.sqrt();
                let new_vac = phi_inv * self.vacuum_amplitude + phi_inv_sqrt * self.tau_amplitude;
                let new_tau = phi_inv_sqrt * self.vacuum_amplitude - phi_inv * self.tau_amplitude;
                self.vacuum_amplitude = new_vac;
                self.tau_amplitude = new_tau;
                self.state = BraidState::RMatrix;
                self.ticks_elapsed = 0;
                false
            }
            BraidState::RMatrix => {
                self.ticks_elapsed += 1;
                self.ring_position = (self.ring_position + 1) % RING_SIZE;
                if self.ticks_elapsed >= TICKS_PER_BRAID {
                    self.vacuum_amplitude *= R_VACUUM_PHASE.cos();
                    self.tau_amplitude *= R_TAU_PHASE.cos();
                    let norm = (self.vacuum_amplitude.powi(2) + self.tau_amplitude.powi(2)).sqrt();
                    if norm > 1e-15 {
                        self.vacuum_amplitude /= norm;
                        self.tau_amplitude /= norm;
                    }
                    self.state = BraidState::FMove;
                    self.ticks_elapsed = 0;
                }
                false
            }
            BraidState::FMove => {
                let phi_inv = 1.0 / PHI;
                let phi_inv_sqrt = phi_inv.sqrt();
                let new_vac = phi_inv * self.vacuum_amplitude + phi_inv_sqrt * self.tau_amplitude;
                let new_tau = phi_inv_sqrt * self.vacuum_amplitude - phi_inv * self.tau_amplitude;
                self.vacuum_amplitude = new_vac;
                self.tau_amplitude = new_tau;
                self.total_braids_executed += 1;
                self.state = BraidState::Complete;
                true
            }
            BraidState::Complete => {
                self.state = BraidState::InverseFMove;
                false
            }
        }
    }

    fn execute_full_braid(&mut self) -> (f64, f64) {
        self.tick(); // F⁻¹
        for _ in 0..TICKS_PER_BRAID { self.tick(); } // R (8 ticks)
        self.tick(); // F
        (self.vacuum_amplitude, self.tau_amplitude)
    }
}

// ── Execute 10 braids and trace the state evolution ──────────────

println!("=== Braiding State Machine: B = F⁻¹ R F ===");
println!("R-matrix phases: R¹ = e^{{-4πi/5}} = {:.4}, Rτ = e^{{3πi/5}} = {:.4}",
    R_VACUUM_PHASE, R_TAU_PHASE);
println!("Tick resolution: 20 Hz (50ms), {} ticks per braid = 500ms/cycle",
    1 + TICKS_PER_BRAID + 1);
println!();

let mut braid = BraidOperator::at_viviani();
println!("Initial state (Viviani peak):");
println!("  |1⟩ = {:.6}, |τ⟩ = {:.6}", braid.vacuum_amplitude, braid.tau_amplitude);
println!("  P(1) = {:.6}, P(τ) = {:.6}",
    braid.vacuum_amplitude.powi(2), braid.tau_amplitude.powi(2));
println!();

println!("{:>5} {:>12} {:>12} {:>12} {:>5}", "Braid", "|1⟩", "|τ⟩", "Norm", "Ring");
println!("{}", "-".repeat(52));

for i in 1..=10 {
    let (vac, tau) = braid.execute_full_braid();
    let norm = vac.powi(2) + tau.powi(2);
    println!("{:5} {:12.6} {:12.6} {:12.8} {:5}",
        i, vac, tau, norm, braid.ring_position);
    braid.tick(); // reset to InverseFMove for next cycle
}

println!();
println!("Total braids executed: {}", braid.total_braids_executed);
````

## 4. Levin-Wen String-Net Operators

The SpiralSafe coherence engine implements the Hamiltonian:

**H = −Σ Qᵥ − Σ Bₚ**

- **Qᵥ (Vertex Operators):** Branching rule enforcement — no orphan τ string endpoints
- **Bₚ (Plaquette Operators):** Charge measurement via Bₚ = (1/D²)(1 + φ·Bₚᵗ)

Where D = √(1 + φ²) ≈ 1.902 is the total quantum dimension of the Doubled Fibonacci model.

The F-matrix for basis changes:

```
F = | φ⁻¹    φ⁻¹ᐟ² |
    | φ⁻¹ᐟ² -φ⁻¹   |
```

This is real, unitary, and self-inverse up to sign — verified below.

````python
// ── Levin-Wen String-Net Operators ────────────────────────────────

const D_SQUARED: f64 = 3.618033988749895; // 1 + φ²

#[derive(Debug)]
struct VertexState { edges: [bool; 3] }

impl VertexState {
    fn qv_eigenvalue(&self) -> f64 {
        let tau_count = self.edges.iter().filter(|&&e| e).count();
        match tau_count {
            0 | 2 | 3 => 1.0,  // Valid branching
            1 => 0.0,           // Orphan τ — violation
            _ => unreachable!(),
        }
    }
}

#[derive(Debug)]
struct PlaquetteState { edge_labels: Vec<bool>, sides: usize }

impl PlaquetteState {
    fn bp_eigenvalue(&self) -> f64 {
        let tau_count = self.edge_labels.iter().filter(|&&e| e).count();
        if tau_count == 0 { 1.0 }
        else { (1.0 + PHI * (tau_count as f64 / self.sides as f64)) / D_SQUARED }
    }
}

// F-matrix unitarity check
fn f_matrix_is_unitary() -> bool {
    let phi_inv = 1.0 / PHI;
    let phi_inv_sqrt = phi_inv.sqrt();
    let col1_norm = phi_inv.powi(2) + phi_inv_sqrt.powi(2);
    let col2_norm = phi_inv_sqrt.powi(2) + phi_inv.powi(2);
    let cross = phi_inv * phi_inv_sqrt + phi_inv_sqrt * (-phi_inv);
    (col1_norm - 1.0).abs() < 1e-10
        && (col2_norm - 1.0).abs() < 1e-10
        && cross.abs() < 1e-10
}

fn f_move(vac: f64, tau: f64) -> (f64, f64) {
    let phi_inv = 1.0 / PHI;
    let phi_inv_sqrt = phi_inv.sqrt();
    (phi_inv * vac + phi_inv_sqrt * tau, phi_inv_sqrt * vac - phi_inv * tau)
}

// ── Demonstrate ───────────────────────────────────────────────────

println!("=== Levin-Wen String-Net Operators ===");
println!("D² = 1 + φ² = {:.6}", D_SQUARED);
println!("D  = √(1+φ²) = {:.6}", D_SQUARED.sqrt());
println!();

// Vertex operator tests
println!("Qv (Vertex Operator) — Branching Rules:");
let configs = [
    ([false, false, false], "(0,0,0) all vacuum"),
    ([true, true, false],   "(τ,τ,0) two τ"),
    ([true, true, true],    "(τ,τ,τ) three τ"),
    ([true, false, false],  "(τ,0,0) orphan τ — VIOLATION"),
];
for (edges, label) in &configs {
    let v = VertexState { edges: *edges };
    let eig = v.qv_eigenvalue();
    let status = if eig == 1.0 { "✓ valid" } else { "✗ REJECTED" };
    println!("  {} → Qv = {} [{}]", label, eig, status);
}

println!();

// Plaquette operator
println!("Bp (Plaquette Operator):");
let vacuum_plaq = PlaquetteState { edge_labels: vec![false, false, false], sides: 3 };
let charged_plaq = PlaquetteState { edge_labels: vec![true, true, false], sides: 3 };
println!("  Vacuum plaquette: Bp = {:.6}", vacuum_plaq.bp_eigenvalue());
println!("  Charged plaquette (2τ/3): Bp = {:.6}", charged_plaq.bp_eigenvalue());

println!();

// F-matrix unitarity
println!("F-Matrix Unitarity: {}", if f_matrix_is_unitary() { "✓ F†F = I" } else { "✗ FAILED" });

// F-move norm preservation
let (a, b) = (1.0 / PHI, (1.0 - 1.0 / (PHI * PHI)).sqrt());
let (a2, b2) = f_move(a, b);
let norm_before = a.powi(2) + b.powi(2);
let norm_after = a2.powi(2) + b2.powi(2);
println!("F-move norm: {:.10} → {:.10} (Δ = {:.2e})",
    norm_before, norm_after, (norm_after - norm_before).abs());
````

## 5. WAVE Protocol — Coherence Scoring

Every piece of content in the system gets scored by four Fibonacci-weighted components:

- **W (Curl):** Circular reasoning detection. Measures lexical repetition — high curl means the text loops on itself.
- **A (Divergence):** Conceptual expansion. Average word length as a proxy for vocabulary breadth.
- **V (Potential):** Structural capacity. Sentence length variance — good prose has rhythm.
- **E (Entropy):** Shannon information density across the byte distribution.

Unified score: C = Σ(fib_weight × component) / Σ(fib_weight), threshold 0.85.

````python
// ── WAVE Protocol Scoring ────────────────────────────────────────

use std::collections::HashSet;

struct WaveScorer { threshold: f64 }

#[derive(Debug)]
struct WaveScore {
    curl: f64, divergence: f64, potential: f64, entropy: f64,
    coherence: f64, fibonacci_weighted: f64, passes: bool,
}

impl WaveScorer {
    fn new() -> Self { Self { threshold: 0.85 } }

    fn score(&self, text: &str) -> WaveScore {
        let curl = self.measure_curl(text);
        let divergence = self.measure_divergence(text);
        let potential = self.measure_potential(text);
        let entropy = self.measure_entropy(text);

        // Fibonacci weights: 1, 1, 2, 3 (sum = 7)
        let coherence = (1.0 * (1.0 - curl) + 1.0 * divergence
            + 2.0 * potential + 3.0 * entropy) / 7.0;

        // φ-weighted variant
        let phi3 = PHI.powi(3);
        let phi2 = PHI.powi(2);
        let fibonacci_weighted = (entropy * phi3 + potential * phi2
            + divergence * PHI + (1.0 - curl)) / (phi3 + phi2 + PHI + 1.0);

        WaveScore {
            curl, divergence, potential, entropy,
            coherence, fibonacci_weighted,
            passes: coherence >= self.threshold,
        }
    }

    fn measure_curl(&self, text: &str) -> f64 {
        let words: Vec<&str> = text.split_whitespace().collect();
        if words.len() < 2 { return 0.0; }
        let unique: HashSet<&&str> = words.iter().collect();
        (1.0 - unique.len() as f64 / words.len() as f64).max(0.0).min(1.0)
    }

    fn measure_divergence(&self, text: &str) -> f64 {
        let words: Vec<&str> = text.split_whitespace().collect();
        if words.is_empty() { return 0.0; }
        let avg = words.iter().map(|w| w.len() as f64).sum::<f64>() / words.len() as f64;
        (avg / 6.0).min(1.0)
    }

    fn measure_potential(&self, text: &str) -> f64 {
        let sents: Vec<&str> = text.split(['.', '!', '?'])
            .filter(|s| !s.trim().is_empty()).collect();
        if sents.is_empty() { return 0.0; }
        let lens: Vec<f64> = sents.iter().map(|s| s.split_whitespace().count() as f64).collect();
        let mean = lens.iter().sum::<f64>() / lens.len() as f64;
        let var = lens.iter().map(|l| (l - mean).powi(2)).sum::<f64>() / lens.len() as f64;
        (var.sqrt() / 10.0).min(1.0)
    }

    fn measure_entropy(&self, text: &str) -> f64 {
        if text.is_empty() { return 0.0; }
        let mut freq = [0u32; 256];
        let total = text.len() as f64;
        for b in text.bytes() { freq[b as usize] += 1; }
        let ent: f64 = freq.iter().filter(|&&f| f > 0)
            .map(|&f| { let p = f as f64 / total; -p * p.log2() }).sum();
        (ent / 6.6).min(1.0)
    }
}

// ── Demonstrate ───────────────────────────────────────────────────

let scorer = WaveScorer::new();

let samples = [
    ("Technical prose",
     "The topological braiding engine processes 9P transactions through CSP channels. \
      Each file excitation carries gauge metadata ensuring conservation. \
      The Fibonacci weighting amplifies coherent signals while attenuating noise."),
    ("Repetitive (high curl)",
     "the the the the the the the the the the the the"),
    ("This notebook application",
     "This Rust EvCxR notebook demonstrates a topological braiding engine built for \
      multi-agent AI coherence. It implements SU(2) level 3 Fibonacci anyons with a \
      conservation law enforced as a hard gauge constraint. The system uses CSP-only \
      concurrency via mpsc channels, eliminating all shared mutable state."),
];

println!("=== WAVE Protocol Scoring ===");
println!();

for (label, text) in &samples {
    let s = scorer.score(text);
    println!("--- {} ---", label);
    println!("  W (curl):       {:.4}", s.curl);
    println!("  A (divergence): {:.4}", s.divergence);
    println!("  V (potential):  {:.4}", s.potential);
    println!("  E (entropy):    {:.4}", s.entropy);
    println!("  Coherence:      {:.4} {}", s.coherence,
        if s.passes { "✓ PASS" } else { "✗ FAIL" });
    println!("  φ-weighted:     {:.4}", s.fibonacci_weighted);
    println!();
}
````

## 6. 9P/Styx Concurrency — CSP-Only (Zero Mutex)

This is the Limbo reincarnation. The concurrency model uses **only** `mpsc` channels — no `Mutex`, no `RwLock`, no shared mutable state. This is an architectural invariant lifted from Bell Labs' Plan 9 / Inferno / Limbo design.

Each 9P file operation becomes a `FileExcitation` message routed through typed channels. The conservation law is checked at the gate — gauge violations are rejected before any state mutation occurs.

The type system below runs synchronously to demonstrate the message types and conservation enforcement. The full async version uses `tokio::sync::mpsc` with the same API.

````python
// ── 9P/Styx File Excitations (sync demonstration) ────────────────

#[derive(Debug, Clone)]
enum NinePOp {
    Tattach { fid: u32, uname: String, aname: String },
    Twalk   { fid: u32, newfid: u32, wnames: Vec<String> },
    Tread   { fid: u32, offset: u64, count: u32 },
    Twrite  { fid: u32, offset: u64, data: Vec<u8> },
    Tclunk  { fid: u32 },
}

#[derive(Debug)]
struct FileExcitation {
    tag: u16,
    op: NinePOp,
    source_strand: String,
    alpha: u32,
    omega: u32,
}

impl FileExcitation {
    fn is_conserved(&self) -> bool { self.alpha + self.omega == 15 }
}

// ── Simulate multi-strand transaction flow ──────────────────────

let transactions = vec![
    FileExcitation {
        tag: 1,
        op: NinePOp::Tattach { fid: 0, uname: "claude".into(), aname: "/coherence".into() },
        source_strand: "claude-reason".into(),
        alpha: 7, omega: 8,
    },
    FileExcitation {
        tag: 2,
        op: NinePOp::Twalk { fid: 0, newfid: 1, wnames: vec!["api".into(), "wave".into()] },
        source_strand: "grok-pulse".into(),
        alpha: 8, omega: 7,
    },
    FileExcitation {
        tag: 3,
        op: NinePOp::Tread { fid: 1, offset: 0, count: 4096 },
        source_strand: "gemini-scale".into(),
        alpha: 6, omega: 9,
    },
    FileExcitation {
        tag: 4,
        op: NinePOp::Twrite { fid: 1, offset: 0, data: b"coherence payload".to_vec() },
        source_strand: "grok-pulse".into(),
        alpha: 7, omega: 7, // VIOLATION: 14 ≠ 15
    },
    FileExcitation {
        tag: 5,
        op: NinePOp::Tclunk { fid: 1 },
        source_strand: "claude-reason".into(),
        alpha: 10, omega: 5,
    },
];

println!("=== 9P/Styx Transaction Processing ===");
println!("Concurrency: CSP-only (mpsc channels, zero Mutex/RwLock)");
println!();

let mut accepted = 0u32;
let mut rejected = 0u32;

for tx in &transactions {
    let conserved = tx.is_conserved();
    let status = if conserved { accepted += 1; "✓ ACCEPTED" }
                 else { rejected += 1; "✗ GAUGE VIOLATION" };
    println!("  [tag={}] {:14} {:?}",
        tx.tag, tx.source_strand, tx.op);
    println!("           α={}, ω={}, sum={} → {}",
        tx.alpha, tx.omega, tx.alpha + tx.omega, status);
    println!();
}

println!("Results: {} accepted, {} rejected", accepted, rejected);
println!("Conservation enforcement rate: {:.0}%",
    100.0 * accepted as f64 / transactions.len() as f64);
````
