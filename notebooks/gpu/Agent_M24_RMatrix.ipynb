## Core R-Matrix (Fundamental U_q(sl_2))

```python
use nalgebra::{Matrix4, Complex};
type C = Complex<f64>;

/// Fundamental R-matrix for the wave coherence model (U_q(sl_2)-style)
fn fundamental_r_matrix(q: f64) -> Matrix4<C> {
    let q_val = C::new(q, 0.0);
    let q_inv = C::new(1.0 / q, 0.0);
    let off = C::new(1.0 - q * q, 0.0);
    Matrix4::from_row_slice(&[
        q_val, C::new(0.0, 0.0), C::new(0.0, 0.0), C::new(0.0, 0.0),
        C::new(0.0, 0.0), q_inv, off, C::new(0.0, 0.0),
        C::new(0.0, 0.0), C::new(0.0, 0.0), q_val, C::new(0.0, 0.0),
        C::new(0.0, 0.0), C::new(0.0, 0.0), C::new(0.0, 0.0), q_inv,
    ])
}

let q = 2.0_f64.sqrt();
let r = fundamental_r_matrix(q);
println!("Fundamental R-matrix for q={:.4}:\n{:?}", q, r);
```
